Importing the dependencies

```python
import pandas as pd
```

Importing the data

```python
train_data = pd.read_csv(r'Datasets/train.csv')
train_data.head()
```

```python
test_data = pd.read_csv(r'Datasets/test.csv')
test_data
```

```python
import re
def text_cleaning(text):
    text = text.lower()
    text = text.replace("\\/", "/").replace("\\n", " ").replace("\t", " ")
    text = re.sub(r"[^a-zA-Z0-9\s.,!?()]+", "", text) 
    text = re.sub(r"\s+", " ", text)
    text = text.strip()
    return text

train_data['prompt'] = train_data['prompt'].apply(text_cleaning)
train_data['response_a'] = train_data['response_a'].apply(text_cleaning)
train_data['response_b'] = train_data['response_b'].apply(text_cleaning)
train_data.head()
```

```python
test_data['prompt'] = test_data['prompt'].apply(text_cleaning)
test_data['response_a'] = test_data['response_a'].apply(text_cleaning)
test_data['response_b'] = test_data['response_b'].apply(text_cleaning)
test_data.head()
```

```python
train_data[['winner_model_a', 'winner_model_b', 'winner_tie']].sum()
```

```python
train_data[['model_a', 'model_b']].nunique()
```

```python
train_data['response_a_len'] = train_data['response_a'].apply(lambda x: len(x))
train_data['response_b_len'] = train_data['response_b'].apply(lambda x: len(x))
```

```python
train_data.head()
```

```python
test_data['response_a_len'] = test_data['response_a'].apply(lambda x: len(x))
test_data['response_b_len'] = test_data['response_b'].apply(lambda x: len(x))
```

```python
test_data.head()
```

```python
train_data['response_a_un'] = train_data['response_a'].apply(lambda x: len(set(x.split(" "))))
train_data['response_b_un'] = train_data['response_b'].apply(lambda x: len(set(x.split(" "))))
train_data['prompt_un'] = train_data['prompt'].apply(lambda x: len(set(x.split(" "))))
```

```python
train_data.head()
```

```python
test_data['response_a_un'] = test_data['response_a'].apply(lambda x: len(set(x.split(" "))))
test_data['response_b_un'] = test_data['response_b'].apply(lambda x: len(set(x.split(" "))))
test_data['prompt_un'] = test_data['prompt'].apply(lambda x: len(set(x.split(" "))))
```

```python
test_data.head()
```

```python
def combine_p_r(row): 
    return (
        f"{row['prompt']}: {row['response_a']}",
        f"{row['prompt']}: {row['response_b']}"
    )
```

```python
train_data[['combined_a', 'combined_b']] = train_data.apply(combine_p_r, axis=1, result_type='expand')
train_data.head()
```

```python
test_data[['combined_a', 'combined_b']] = test_data.apply(combine_p_r, axis=1, result_type='expand')
test_data.head()
```

```python
train_data['winner'] = pd.from_dummies(train_data[['winner_model_a', 'winner_model_b', 'winner_tie']])
train_data['winner'] = train_data['winner'].map({'winner_model_a':0, 'winner_model_b':1, 'winner_tie':2})
train_data.drop(['winner_model_a', 'winner_model_b', 'winner_tie'], axis=1, inplace=True)
train_data.head()
```

```python
X = train_data[["prompt", "response_a", "response_b", "response_a_len", "response_b_len", "response_a_un", "response_b_un", 
              "prompt_un", "combined_a", "combined_b"]]
Y = train_data[['winner']]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from scipy.sparse import hstack
from sklearn.metrics import accuracy_score
```

```python
numeric_features = [
    'response_a_len', 'response_b_len', 'response_a_un', 'response_b_un', 'prompt_un'
]

text_features = [
    'prompt', 'response_a', 'response_b', 'combined_a', 'combined_b'
]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# TF-IDF Vectorization
tfidf_prompt = TfidfVectorizer(max_features=1500, stop_words='english')
tfidf_responses = TfidfVectorizer(max_features=1500, stop_words='english')

prompt_tf = tfidf_prompt.fit_transform(X_train['prompt'])
response_a_tf = tfidf_responses.fit_transform(X_train['response_a'])
response_b_tf = tfidf_responses.fit_transform(X_train['response_b'])
combined_a_tf = tfidf_responses.fit_transform(X_train['combined_a'])
combined_b_tf = tfidf_responses.fit_transform(X_train['combined_b'])

# Scaling numeric features
scaler = StandardScaler()
X_train_numeric = scaler.fit_transform(X_train[numeric_features])
X_test_numeric = scaler.transform(X_test[numeric_features])

# Combine TF-IDF and scaled numeric features
X_train_tf = hstack([
    prompt_tf, 
    response_a_tf, 
    response_b_tf, 
    combined_a_tf, 
    combined_b_tf, 
    X_train_numeric
])

# Do the same for test set
prompt_tf_test = tfidf_prompt.transform(X_test['prompt'])
response_a_tf_test = tfidf_responses.transform(X_test['response_a'])
response_b_tf_test = tfidf_responses.transform(X_test['response_b'])
combined_a_tf_test = tfidf_responses.transform(X_test['combined_a'])
combined_b_tf_test = tfidf_responses.transform(X_test['combined_b'])

X_test_tf = hstack([
    prompt_tf_test, 
    response_a_tf_test, 
    response_b_tf_test, 
    combined_a_tf_test, 
    combined_b_tf_test, 
    X_test_numeric
])
```

```python
from xgboost import XGBClassifier
```

```python
xgb_model = XGBClassifier(n_estimators=500, subsample=0.8, min_child_weight=10, max_depth=3, learning_rate=0.1, gamma=0, colsample_bytree=0.8)
xgb_model.fit(X_train_tf, Y_train)
Y_pred = xgb_model.predict(X_test_tf)
print(f"Accuracy: {accuracy_score(Y_test, Y_pred)}")
```
